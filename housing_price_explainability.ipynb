# Housing Price Prediction & Explainability

Compare Random Forest with mean and linear-regression baselines using California census district data. Median income is in tens of thousands of dollars; room counts refer to a district, not a property.

PDP/ICE and quantile-bin ALE show model associations, not causal effects. The ALE implementation uses within-bin finite differences and observation-weight centering. Evaluation uses a random 80/20 split, not a geographic holdout; no hyperparameter search is claimed.
